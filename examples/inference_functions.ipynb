```python
antoines = predict_vp_parameters(
    smiles="CCO",
    temperature=298.15,
)
display(antoines)

def softplus_beta(x, beta):
    return softplus(beta * x) / beta

predicted_vp = 10 ** (antoines['antoine_a'] - softplus(antoines['antoine_b']) / softplus_beta(298.15 / antoines['antoine_t_scale'] + antoines['antoine_c'], beta=0.1))
display(predicted_vp)
```
